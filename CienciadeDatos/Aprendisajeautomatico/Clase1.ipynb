# Ejercicios de Arrays y Matrices con NumPy

```python
import numpy as np
```

## Ejercicio 1: Crear un array de 10 elementos con ceros

```python
arr_ceros = np.zeros(10)
print("Array de 10 ceros:", arr_ceros)
```

## Ejercicio 2: Genera un array con los números del 1 al 20

```python
arr_1_20 = np.arange(1, 21)
print("Array del 1 al 20:", arr_1_20)
```

## Ejercicio 3: Crea un array con los números pares del 2 al 50

```python
arr_pares = np.arange(2, 51, 2)
print("Números pares del 2 al 50:", arr_pares)
```

## Ejercicio 4: Crea una matriz identidad de 5x5

```python
mat_identidad = np.eye(5)
print("Matriz identidad 5x5:")
print(mat_identidad)
```

## Ejercicio 5: Crea una matriz de 4x4 con valores aleatorios entre 0 y 1

```python
mat_random = np.random.rand(4, 4)
print("Matriz 4x4 aleatoria (0 a 1):")
print(mat_random)
```

## Ejercicio 6: Suma elemento a elemento de a=[1,2,3] y b=[4,5,6]

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
suma = a + b
print(f"a = {a}")
print(f"b = {b}")
print(f"a + b = {suma}")
```

## Ejercicio 7: Multiplica dos matrices 3x3 de enteros aleatorios entre 1 y 10

```python
mat_a = np.random.randint(1, 11, size=(3, 3))
mat_b = np.random.randint(1, 11, size=(3, 3))
producto = mat_a @ mat_b  # equivalente a np.dot(mat_a, mat_b)

print("Matriz A:")
print(mat_a)
print("\nMatriz B:")
print(mat_b)
print("\nA × B:")
print(producto)
```

## Ejercicio 8: Array de 15 valores aleatorios (1-100) → máximo, mínimo y promedio

```python
arr_random = np.random.randint(1, 101, size=15)
print("Array:", arr_random)
print(f"Máximo:  {arr_random.max()}")
print(f"Mínimo:  {arr_random.min()}")
print(f"Promedio: {arr_random.mean():.2f}")
```

## Ejercicio 9: Matriz 5x5 con valores del 1 al 25 → extraer diagonal

```python
mat_5x5 = np.arange(1, 26).reshape(5, 5)
diagonal = np.diag(mat_5x5)

print("Matriz 5x5:")
print(mat_5x5)
print(f"\nDiagonal: {diagonal}")
```

## Ejercicio 10: Matriz 5x5 aleatoria (1-20) → suma de cada fila y columna

```python
mat = np.random.randint(1, 21, size=(5, 5))
suma_filas = mat.sum(axis=1)
suma_columnas = mat.sum(axis=0)

print("Matriz:")
print(mat)
print(f"\nSuma por fila:    {suma_filas}")
print(f"Suma por columna: {suma_columnas}")
```
