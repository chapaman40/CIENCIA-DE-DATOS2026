# Actividad Semana 2 — Importación y Manipulación de Datos con PANDAS

Contexto: Proyecto de análisis de mercado para una empresa de comercio electrónico.  
Los datos provienen de tres fuentes distintas: un archivo CSV, un archivo JSON y una hoja de cálculo Excel.

---

```python
# Importación de la librería PANDAS
import pandas as pd
```

## 1. Carga de datos de ventas desde el archivo CSV

Se importa el archivo `ventas (1).csv` que contiene información sobre las ventas de productos.  
Cada fila representa una venta individual con columnas de id, fecha, producto, cantidad y precio unitario.

```python
# Cargar el archivo CSV
df_ventas = pd.read_csv('archivos semana 2/ventas (1).csv')

# Vista previa de las primeras filas
print('Vista previa de los datos de ventas' )
print(df_ventas.head())
print('Dimensiones del DataFrame:', df_ventas.shape)
print('Tipos de datos por columna:')
print(df_ventas.dtypes)
```

**Interpretación:** La función `head()` muestra las primeras 5 filas del DataFrame. Podemos observar las columnas disponibles y los tipos de datos de cada una. El atributo `shape` indica la cantidad total de filas y columnas.

---
## 2. Carga de datos de clientes desde el archivo JSON

Se importa el archivo `clientes.json` que contiene los datos de los clientes que realizaron compras.  
Cada objeto JSON representa un cliente con su id, nombre, email y teléfono.

```python
# Cargar el archivo JSON
df_clientes = pd.read_json('archivos semana 2/clientes.json')

# Mostrar el DataFrame completo
print('Datos de clientes')
print(df_clientes)
print()

# Calcular y mostrar la cantidad total de clientes
total_clientes = len(df_clientes)
print(f'Cantidad total de clientes: {total_clientes}')
```

**Interpretación:** La función `read_json()` convierte automáticamente el array de objetos JSON en un DataFrame. La función `len()` sobre el DataFrame devuelve la cantidad de filas, equivalente al número total de clientes registrados.

---
## 3. Carga de datos de inventario desde la hoja de cálculo Excel

Se importa el archivo `inventario.xlsx` que contiene información sobre el inventario de productos.  
Se calculará el **promedio de stock disponible** para todos los productos.

```python
# Cargar el archivo Excel
df_inventario = pd.read_excel('archivos semana 2/inventario.xlsx')

# Vista previa del inventario
print('Datos de inventario')
print(df_inventario.head())
print('Columnas disponibles:', df_inventario.columns.tolist())


# Calcular el promedio de stock disponible
# Se busca la columna que contenga 'stock' en su nombre (case-insensitive)
columna_stock = [col for col in df_inventario.columns if 'stock' in col.lower()]

if columna_stock:
    col = columna_stock[0]
    promedio_stock = df_inventario[col].mean()
    print(f'Promedio de stock disponible (columna "{col}"): {promedio_stock:.2f} unidades')
else:
    print('No se encontró una columna de stock. Columnas disponibles:', df_inventario.columns.tolist())
    print('Estadísticas descriptivas del inventario:')
    print(df_inventario.describe())
```

**Interpretación:** La función `read_excel()` importa la hoja de cálculo directamente en un DataFrame. El método `mean()` calcula el promedio aritmético de los valores de stock, lo que nos da una idea del nivel medio de inventario disponible por producto.
